```python
import pandas as pd

df = pd.read_csv("customer_churn.csv")

display(df.head())
display(df.shape)
```

#### Unwanted Column Remove

```python
df = df.drop("customerID", axis=1)

display(df.head())
```

#### Missing Value Handle

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df = df.dropna()
```
